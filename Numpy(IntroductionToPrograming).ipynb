````python
#01
'''
Create a 3x3 array with random integers between 1 and 100. Write a program to
"normalize" the array so that the values are scaled between 0 and 1.
• Formula: (x - min) / (max - min)
• Example Output: An array where the smallest value becomes 0.0 and the
largest becomes 1.0.

'''
import numpy as np

# Create a 3x3 array with random integers between 1 and 100 (101 is exclusive)
arr = np.random.randint(1, 101, size=(3, 3))

# Find the minimum and maximum values in the array
min_val = arr.min()
max_val = arr.max()

# Apply Min-Max normalization: (x - min) / (max - min)
normalized_arr = (arr - min_val) / (max_val - min_val)

print("Original Array:\n", arr)
print("\nNormalized Array:\n", np.round(normalized_arr, 3))
````

````python
#02
''''
2. Create a 1D array of 10 random numbers. Ask the user for a value X Find the value
in the array that is closest to X
• Example Input: Array: [1, 5, 8, 12], X = 10
• Example Output: 8 or 12 (whichever is mathematically closer).
'''
# Create a 1D array of 10 random integers between 1 and 50
arr = np.random.randint(1, 51, size=10)
print("Generated Array:", arr)

# Ask the user for a target value X
x = float(input("Enter a value X: "))

# Calculate the absolute difference between every element and X,
# then find the index of the smallest difference
closest_index = np.abs(arr - x).argmin()
closest_value = arr[closest_index]

print(f"Closest value to {x} is: {closest_value}")
````

````python
#03
''''
3. Write a NumPy program to create a 5x5 array with random values. For each column
in the array, find the second-largest value.
• Example Input: [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
• Example Output: [4, 5, 6] (These are the second-highest values for each
respective column).
'''
# Create a 5x5 array with random integers between 1 and 100
arr = np.random.randint(1, 101, size=(5, 5))

# Sort each column vertically (axis=0) in ascending order,
# then select the second-to-last row [-2]
second_largest = np.sort(arr, axis=0)[-2]

print("Generated 5x5 Array:\n", arr)
print("\nSecond-largest value in each column:", second_largest)
````

````python
#04
'''
4. Create a 6x4 matrix of random integers. Identify which rows have a sum that is an
even number. Print the indices of these rows.
'''
# Create a 6x4 matrix with random integers between 1 and 100
matrix = np.random.randint(1, 101, size=(6, 4))

# Calculate the sum of each row
row_sums = np.sum(matrix, axis=1)

# Find the indices of rows with even sums
even_sum_indices = np.where(row_sums % 2 == 0)[0]

print("Generated 6x4 Matrix:\n", matrix)
print("\nIndices of rows with even sums:", even_sum_indices)
````

````python
#05
'''
5. Create a 1D array of 30 random integers between 1 and 50. Replace all numbers
that are multiples of 3 or 5 with the value -1.
• Example Input: [3, 7, 10, 15]
• Example Output: [-1, 7, -1, -1]
'''
# Create a 1D array of 30 random integers between 1 and 50
arr = np.random.randint(1, 51, size=30)

# Replace multiples of 3 or 5 with -1
arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1
print("Generated 1D Array:\n", arr)
````

````python
#06
''''
6. Write a NumPy program to create a 3x3 array with random values. For each row,
calculate its mean (average) and subtract that mean from every element in that
specific row.
• Example Input: ``` [[10, 20, 30], [5, 5, 5], [0, 10, 20]]
• Example Output: [[-10. 0. 10.], [ 0. 0. 0.], [-10. 0. 10.]]
'''
# Create a 3x3 array with random integers between 1 and 100
arr = np.random.randint(1, 101, size=(3, 3))    
print("Generated 3x3 Array:\n", arr)

# Calculate the mean of each row    
row_means = np.mean(arr, axis=1, keepdims=True)

# Subtract the mean from each element in the respective row
# np.subtract(arr, row_means) is equivalent to arr - row_mean
result = np.subtract(arr, row_means)

print("\nResult after subtracting row means:\n", result)    



````
